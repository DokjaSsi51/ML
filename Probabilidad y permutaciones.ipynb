```python
import numpy as np
import matplotlib.pyplot as plt

sci = lambda Ans: format(Ans, ".5E")

def fact(n):
    return np.prod(range(1, int(n)+1))

def set_value(text = None, value = None):
    value = None
    if value is not None:
        return value
    else:
        return float(input(text))
        
def Factorizer(matrix, vector):
    matrix0 = np.array(matrix, dtype = "float")
    n, m = matrix0.shape
    Q = np.zeros((n,m), dtype = "float")
    R = np.zeros((m,m), dtype = "float")
    u = np.zeros((n,m), dtype = "float")
    e = np.zeros((n,m), dtype = "float")
    x = np.zeros(n, dtype = "float")
    u[:,0] = matrix[:,0]
    e[:,0] = u[:,0] / np.linalg.norm(u[:,0])

    for i in range(m):
        u[:,i] = matrix0[:,i]
        
        for j in range(i):
            u[:,i] -= ((matrix0[:,i]@e[:,j]) * e[:,j])
            
        e[:,i] = u[:,i] / np.linalg.norm(u[:,i])
        
    for i in range(n):
        for j in range(i, m):
            if i <= j:
                R[i,j] = (matrix0[:,j]@e[:,i])
                          
    Q = np.copy(e)
    for i in range(n - 1, -1, -1):
        x[i] = ((Q.T[i,:]@vector) - np.dot(R[i,:], x)) / R[i,i]

    return Q, R, x
```
