# INTERPOLACIÓN #

```python
from sklearn.datasets import make_classification
from imblearn.over_sampling import RandomOverSampler
from imblearn.under_sampling import RandomUnderSampler
from collections import Counter

x, y = make_classification(n_samples = 1000, n_features = 10, n_classes = 2, weights= [0.9, 0.1])
```

```python
np.unique(y)
```

```python
def distribucion(y, titulo):
    counter = Counter(y)
    print(f"{titulo}: {counter}")

distribucion(y, "Distribución de clase")
```

Realizando un sobremuestreo haciendo una interpolación de datos determinando un punto medio entre dos datos. 

```python
over = RandomOverSampler(sampling_strategy = "auto", random_state = 42)
x_over, y_over = over.fit_resample(x,y)
distribucion(y_over, "Distribución de clase después de sobremuestreo")
```

Genera nuevos datos para igualar los datos, esto mediante la interpolación genera datos sintéticos. Los el numero datos de cada clase se iguala a la cantidad de datos que tiene más datos

```python
under = RandomUnderSampler(sampling_strategy = "auto", random_state = 42)
x_under, y_under = under.fit_resample(x,y)
distribucion(y_under, "Distribución de clase después de submmuestreo")
```

Se balancea el dataset de tal manera que cada clase tenga la misma cantidad de datos que la clase con menos datos.

# Support Vectorial Regression #

```python
mojarra = pd.read_csv("Datasets/auto-mpg.csv")
mojarra.head(5)  
```

```python
mojarra = mojarra.dropna(axis = 0, how = "any") #Eliminamos los valores nulos
mojarra.info()
```

```python
from sklearn.svm import SVR


Lista = ["horsepower", "weight"]
x = np.array(mojarra[Lista])
y = np.array(mojarra[["displacement"]])

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2)

svr = SVR()
svr.fit(x_train,y_train)

y_pred = svr.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

print("erre do': ", r2_score(y_test, y_pred))

plt.scatter(y_test, y_pred)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()],"k--")
plt.show()
```

```python
plt.scatter(x_test[:,0], y_pred) #Datos predichos
plt.scatter(x_test[:,0], y_test) #Datos reales
plt.show()
```
